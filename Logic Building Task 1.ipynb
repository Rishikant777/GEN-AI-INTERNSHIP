```python
#User Login Check
a_user = "admin"
a_pwd = "1234"

# user input
user = input("Username: ")
pwd = input("Password: ")

if user == a_user and pwd == a_pwd:
    print("Login Successful")
else:
    print("Invalid Credentials")
```

```python
#Pass / Fail Analyzer
marks = [45, 78, 90, 33, 60]

p = 0
f = 0

for i in marks:
    if i >= 50:
        p += 1
    else:
        f += 1

print("Total:", len(marks))
print("Pass:", p)
print("Fail:", f)
```

```python
# Simple Data Cleaner
names = [" Alice ", "bob", " CHARLIE "]

c = []

for n in names:
    if n:
        c.append(n.strip().lower())
    else:
        c.append("")

print(c)

```

```python
#Message Length Analyzer
messages = ["Hi", "Welcome to the platform", "OK"]

for m in messages:
    l = len(m)
    print(m, "-", l)

    if l > 10:
        print("long")
    
    print()
```

```python
#Error Message Detector
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

c = 0

for i in logs:
    if i == "ERROR":
        c += 1

print(c)
```
