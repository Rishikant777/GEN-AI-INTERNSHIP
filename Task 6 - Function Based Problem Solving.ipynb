```python
#Smart Parking Lot Management System
def smart_parking(capacity, logs):
    current = 0
    peak = 0

    for log in logs:
        if log == "IN":
            current += 1
        elif log == "OUT":
            if current > 0:
                current -= 1
        if current > peak:
            peak = current

    print("Currently Parked Vehicles:", current)
    print("Peak Parking Usage:", peak)

    if current > capacity:
        print("Parking Status: Full - Capacity Exceeded!")
    else:
        print("Parking Status: Available")


# Sample Input
capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking(capacity, vehicle_logs)
```

```python
#Online Food Delivery Time Estimator
def estimate_delivery(distance, traffic, weather):
    speed = 30  
    time = (distance / speed) * 60
    if traffic == "High":
        time += 20
    elif traffic == "Medium":
        time += 10
    else:
        time += 0
    if weather == "Rainy":
        time += 15
    elif weather == "Stormy":
        time += 25
    else:
        time += 0

    print("Estimated Delivery Time:", int(time), "minutes")

# Sample Input
distance_km = 8
traffic_level = "High"
weather_condition = "Rainy"

estimate_delivery(distance_km, traffic_level, weather_condition)
```

```python
#Movie Theatre Seat Occupancy
def seat_analyzer(total_seats, booked_list):
    booked_count = 0
    for seat in booked_list:
        if seat == 1:
            booked_count += 1
    occupancy = (booked_count / total_seats) * 100

    print("Occupancy:", int(occupancy), "%")

    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
        print("Suggestion: You can open additional show")
    else:
        print("Show Status: Seats Available")


# Sample Input
total = 200

# creating 150 booked manually
booked = []
for i in range(150):
    booked.append(1)

seat_analyzer(total, booked)
```

```python
#Cloud Server Load Classification System
def check_server_load(cpu_list):
    total = 0
    for value in cpu_list:
        total += value

    average = total / len(cpu_list)
    print("Average CPU Load:", int(average), "%")
    if average < 50:
        print("Server Status: Normal")
    elif average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# Sample Input
cpu_readings = [45, 60, 70, 85, 90]

check_server_load(cpu_readings)
```

```python
#Smart Classroom Resource Usage Monitor
def resource_tracker(usage_data):
    overused = []

    for item in usage_data:
        hours = usage_data[item]
        if hours > 8:
            overused.append(item)
    if len(overused) > 0:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Input
resources = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

resource_tracker(resources)
```
