```python
# Unique Words in a Sentence

a = input("Enter sentence: ")

# break sentence into words
b = a.split()

unique = []

for w in b:
    if w not in unique:
        unique.append(w)

print("Unique words count:", len(unique))
print("Unique words:", set(unique))
```

```python
# Highest Salary from Employee Data
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

highest = 0
person = None

for i in employees:
    sal = employees[i]
    if sal > highest:
        highest = sal
        person = i

print("Highest Salary:", person, "-", highest)
```

```python
#Find Maximum and Minimum Values
numbers = [45, 22, 89, 10, 66]

# take first value as starting max and min
max_val = numbers[0]
min_val = numbers[0]

for n in numbers:
    if n > max_val:
        max_val = n
    if n < min_val:
        min_val = n

print("List:", numbers)
print("Max:", max_val)
print("Min:", min_val)
```

```python
#Count Products Above a Price Threshold

prices = [450, 1200, 899, 1500, 300]
countes = 0

for p in prices:
    if p > 1000:
        countes += 1

print("Products above 1000:", countes)
```

```python
#Calculate Attendance Percentage
attendance = ["P", "P", "A", "P", "P"]

p = 0

for i in range(len(attendance)):
    if attendance[i] == "P":
        p = p + 1

per = p * 100 / len(attendance)

print("Attendance Percentage:", per)
```

```python
#Remove Duplicate Phone Numbers
phone_numbers = [9876543210, 9123456789, 9876543210]

unique = []

for num in phone_numbers:
    if num not in unique:
        unique.append(num)

print("Unique phone numbers:", set(unique))
```

```python
# Count Character Frequency
s = "pythonp"

d = {}

for i in range(len(s)):
    c = s[i]

    if c not in d:
        d[c] = 1
    else:
        d[c] = d[c] + 1

print(d)
```

```python
#Check if a Key Exists in a Dictionary
data = {
    "Employee": "Ravi",
    "Salary": 50000,
    "Dept": "IT"
}

key = "Employee"

if key in data:
    print("Employee exists")
else:
    print("Employee not found")
```

```python
#Calculate Average Marks
marks = [70, 75, 80, 65]

total = 0

for m in marks:
    total += m

avg = total / len(marks)

print("Average Marks:", avg)
```
