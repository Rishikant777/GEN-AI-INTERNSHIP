```python
#Social Media – Post Engagement Analyzer
def analyze_post(likes):
    total = 0
    for i in likes:
        total = total + i
    print("Total Likes:", total)
    if total >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")


# taking sample data
likes_list = [200, 150, 300, 400]

analyze_post(likes_list)
```

```python
#Healthcare – Medicine Stock Alert System
def check_eligibility(age):
    if age >= 18:
        return "Eligible"
    else:
        return "Not Eligible"


# Taking input from user
age = int(input("Enter Patient Age: "))

status = check_eligibility(age)

print("Patient Age:", age)
print("Eligibility Status:", status)
```

```python
# Agriculture – Premium Crop Price Filter
def filter_premium_crops(price_list):
    premium = []

    for price in price_list:
        if price > 2000:
            premium.append(price)

    return premium
crop_prices = [1500, 2500, 1800, 3200, 2000]

premium_crops = filter_premium_crops(crop_prices)

print("Premium Crops:", premium_crops)
```

```python
#System Monitoring – Application Health 
def system_status(errors):
    if errors == 0:
        return "Healthy"
    elif errors <= 5:
        return "Minor Issues"
    else:
        return "Critical Issues"


errors = int(input("Enter Error Count: "))

print("Error Count:", errors)
print("System Status:", system_status(errors))
```

```python
#Banking – Daily Transaction Limit Checker
def check_transaction(amount):
    limit = 50000

    if amount <= limit:
        return "Approved"
    else:
        return "Rejected"


amt = int(input("Enter Transaction Amount: "))
result = check_transaction(amt)
print("Transaction Amount:", amt)
print("Transaction Status:", result)
```

```python
# E-Learning – Student Attendance Eligibility System
def check_exam_eligibility(days_present, total_days):
    count = 0

    for day in days_present:
        if day == 1:
            count += 1

    percentage = (count / total_days) * 100

    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    return percentage, status

attendance = [1, 1, 0, 1, 1]
total_classes = len(attendance)

percent, result = check_exam_eligibility(attendance, total_classes)

print("Attendance Percentage:", percent)
print("Exam Eligibility:", result)
```

```python
#Smart Electricity Bill Analyzer
def electricity_bill(units):
    bill = 0

    if units <= 100:
        bill = units * 3
    else:
        bill = 100 * 3
        if units <= 200:
            bill = bill + (units - 100) * 5
        else:
            bill = bill + 100 * 5
            bill = bill + (units - 200) * 7

    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    return bill, usage


u = int(input("Enter Units: "))

total, level = electricity_bill(u)

print("Total Bill:", total)
print("Usage Level:", level)
```
