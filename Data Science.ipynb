```python
# Mobile Recharge Validation System
def check_recharge(amount):
    plans = [199, 299, 399, 599]   
    if amount < 50:
        print("Minimum recharge is ₹50")
        return False    
    if amount not in plans:
        print("Invalid plan. Choose from 199, 299, 399, 599")
        return False    
    print("Recharge Done Successfully!")
    return True
# retry until valid
while True:
    amt = int(input("Enter recharge amount: "))
    if check_recharge(amt):
        break
    else:
        print("Try Again\n")
```

```python
# Inventory Reorder Alert System
def stock_check(data):
    for item, qty in data.items():
        if qty < 15:
            print(item + " : Reorder Alert")
        else:
            print(item + " : Stock OK")
products = {
    "Rice": 20,
    "Sugar": 12,
    "Oil": 30,
    "Soap": 5
}

stock_check(products)
```

```python
# Student Result Processing System
def check_result(marks):
    total = 0    
    for i in marks:
        total += i    
    avg = total / len(marks)
    print("Average:", avg)    
    if avg >= 50:
        return "Pass"
    else:
        return "Fail"
# input
nums = [40, 60, 55, 45, 70]
res = check_result(nums)
print("Result:", res)
```

```python
#Cab Fare Estimator with Retry Option
def fare_calc(km, peak):
    fare = 50 + km * 12
    if peak == "y":
        fare = fare + fare * 0.25
    return fare

while True:
    d = float(input("Enter distance (km): "))
    p = input("Peak hour? (y/n): ")
    amount = fare_calc(d, p)
    print("Total Fare =", amount)
    ch = input("Calculate again? (y/n): ")
    if ch == "n":
        break

    print()
```

```python
#Employee Attendance Eligibility 
def attendance_check(data):
    total = len(data)
    p = 0
    for i in data:
        if i == "P":
            p += 1
    per = (p / total) * 100
    print("Percentage:", per)

    if per >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

# sample list
att = ["P", "P", "A", "P", "A", "P", "P"]

result = attendance_check(att)
print("Result:", result)
```

```python
#Password Strength Checker
def password_check(p):
    if len(p) < 8:
        return "Weak Password"
    d = 0
    s = 0
    for i in p:
        if i.isdigit():
            d += 1
        if i == "@" or i == "#" or i == "$":
            s += 1
    if d > 0 and s > 0:
        return "Strong Password"
    else:
        return "Weak Password"
pwd = input("Enter password: ")
print(password_check(pwd))
```
