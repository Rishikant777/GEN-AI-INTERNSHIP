```python
#Smart Light Controller
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"  

    def turn_on(self):
        self.status = "ON"

    def turn_off(self):
        self.status = "OFF"

    def show_status(self):
        print(self.name + " is " + self.status)

light_name = input("Light Name: ")
action = input("Action: ")
light = SmartLight(light_name)
if action == "ON":
    light.turn_on()
elif action == "OFF":
    light.turn_off()
light.show_status()
```

```python
#Employee ID Card System
class Employee:
    def __init__(self, name, emp_id, dept):
        self.name = name
        self.emp_id = emp_id
        self.dept = dept

    def show_id_card(self):
        print("Employee ID Card")
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.dept)

name = input("Employee Name: ")
emp_id = input("Employee ID: ")
dept = input("Department: ")
emp1 = Employee(name, emp_id, dept)
emp1.show_id_card()
```

```python
#Mobile Contact Record
class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def show_contact(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.phone)

name = input("Contact Name: ")
phone = input("Phone Number: ")
c1 = Contact(name, phone)
c1.show_contact()
```

```python
#Product Price Tag Generator
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def show_price_tag(self):
        print("Product:", self.name)
        print("Price: ₹" + str(self.price))

name = input("Product Name: ")
price = input("Price: ")
p1 = Product(name, price)
p1.show_price_tag()
```

```python
#Movie Rating Display System
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def show_details(self):
        print("Movie:", self.name)
        print("Rating:", self.rating, "/ 5")
        
name = input("Movie Name: ")
rating = input("Rating: ")
m1 = Movie(name, rating)
m1.show_details()
```

```python
# Delivery Address Manager
class Customer:
    def __init__(self, name, address):
        self.name = name
        self.address = address

    def show_delivery(self):
        print("Delivery Details")
        print("Customer:", self.name)
        print("Address:", self.address)

name = input("Customer Name: ")
address = input("Address: ")
c1 = Customer(name, address)
c1.show_delivery()
```
